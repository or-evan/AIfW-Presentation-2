In order to use eumdac data, create a file named Keys.csv and add it to your .gitignore with the format

consumer_key,consumer_secret

```python
import csv
import datetime
import os
with open('Keys.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        consumer_key = row[0]
        consumer_secret = row[1]

```

```python
import eumdac


# Insert your personal key and secret into the single quotes
credentials = (consumer_key, consumer_secret)

token = eumdac.AccessToken(credentials)

print(f"This token '{token}' expires {token.expiration}")
```

```python
datastore = eumdac.DataStore(token)

collection_id = "EO:EUM:DAT:MSG:HRSEVIRI"
collection = datastore.get_collection(collection_id)
print(f"Collection Title: {collection.title}")

# 2. Define time window
start_time = datetime.datetime(2026, 8, 27, 7, 0)
end_time = datetime.datetime(2026, 8, 29, 7, 0)

products = list(collection.search(dtstart=start_time, dtend=end_time))

if not products:
    print("No products found for the given criteria.")
else:
    target_product = products[0]
    print(f"Found product: {target_product}")

    output_dir = "./msg_seviri_data"
    os.makedirs(output_dir, exist_ok=True)

    # 3. Download all files in the product (without filtering for .nc)
    downloaded_files = []
    for entry in target_product.entries:
        out_path = os.path.join(output_dir, str(entry))
        print(f"Downloading {entry}...")
        with target_product.open(entry) as f_in, open(out_path, "wb") as f_out:
            f_out.write(f_in.read())
        downloaded_files.append(out_path)

    print(
        f"\nSuccessfully downloaded {len(downloaded_files)} files to {output_dir}"
    )
```
