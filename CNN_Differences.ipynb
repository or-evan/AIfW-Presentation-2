```python
import csv
key_map = {}

with open('Keys.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        if len(row) >= 2:
            # row[0] is the label ('key1'), row[1] is the value ('ACTUALKEY1')
            key_map[row[0].strip()] = row[1].strip()

actual_key1 = key_map.get('consumer_key')
actual_key2 = key_map.get('consumer_secret')

print(actual_key1)  # Output: ACTUALKEY1
print(actual_key2)  # Output: ACTUALKEY2
```
